### Using Python, Pandas, Matplotlib and Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
churn_data = pd.read_csv("churn.csv")

churn_data.head()
```

```python
churn_data.info()
```

```python
churn_data.isnull().sum()
```

```python
churn_data = churn_data.drop_duplicates()

churn_data.shape
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    churn_data['Age'],
    bins=20,
    kde=True
)

plt.title('Age Distribution')

plt.xlabel('Age')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    churn_data['CreditScore'],
    bins=20,
    kde=True
)

plt.title('Credit Score Distribution')

plt.xlabel('Credit Score')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    churn_data['Balance'],
    bins=20,
    kde=True
)

plt.title('Balance Distribution')

plt.xlabel('Balance')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Tenure',
    data=churn_data
)

plt.title('Customer Tenure Distribution')

plt.xlabel('Tenure')

plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='Exited',
    data=churn_data
)

plt.title('Customer Churn Count')

plt.xlabel('Exited')

plt.ylabel('Count')

plt.show()
```

```python
numeric_data = churn_data.select_dtypes(
    include=['int64', 'float64']
)

correlation = numeric_data.corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='Exited',
    y='Age',
    data=churn_data
)

plt.title('Churn by Age')

plt.xlabel('Exited')

plt.ylabel('Age')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='Exited',
    y='Balance',
    data=churn_data
)

plt.title('Churn by Balance')

plt.xlabel('Exited')

plt.ylabel('Balance')

plt.show()
```

```python
average_values = churn_data.groupby(
    'Exited'
)[
    ['Age', 'Balance', 'CreditScore']
].mean()

average_values
```

# Insights

- Customers with higher age showed higher churn tendency.
- Customers with higher account balances were more likely to leave the bank.
- Credit score had a weaker relationship with churn compared to age and balance.
- Certain tenure groups showed higher churn patterns.
- Correlation analysis helped identify important customer behavior trends.
